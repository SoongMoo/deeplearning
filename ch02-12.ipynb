# 선형 회귀

```python
import numpy as np
import matplotlib.pyplot as plt

#텐서플로의 케라스 API에서 필요한 함수들을 불러 옵니다.
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

x = np.array([2, 4, 6, 8])
y = np.array([81, 93, 91, 97])


model = Sequential()
model.add(Dense(1, input_dim=1, activation='linear'))


model.compile(optimizer='sgd', loss='mse')
model.fit(x, y, epochs=500, verbose=0)
plt.scatter(x, y)
plt.plot(x, model.predict(x),'r')
plt.show()
```

```python
import tensorflow as tf
#임의의 학습 시간과 과외 시간을 집어넣어 점수를 예측하는 모델을 테스트해 보겠습니다.

hour = 6
input_data = tf.constant([hour])
prediction = model.predict(input_data)[0][0]
print("%.f시간을 공부한 경우, 예상 점수는 %.02f점입니다" % (hour, prediction))
```

# 다중 선형회귀

```python
import numpy as np
import matplotlib.pyplot as plt

#텐서플로의 케라스 API에서 필요한 함수들을 불러 옵니다.
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense


x = np.array([[2, 0], [4, 4], [6, 2], [8, 3]])
y = np.array([81, 93, 91, 97])


model = Sequential()

model.add(Dense(1, input_dim=2, activation='linear' ))
```

```python
model.compile(optimizer='sgd' ,loss='mse')

model.fit(x, y, epochs=500, verbose=0)
```

```python
import tensorflow as tf
#임의의 학습 시간과 과외 시간을 집어넣어 점수를 예측하는 모델을 테스트해 보겠습니다.

hour = 6
private_class = 3

input_data = tf.constant([[hour, private_class]])
prediction = model.predict(input_data)[0][0]

print("%.f시간을 공부하고 %.f시간의 과외를 받을 경우, 예상 점수는 %.02f점입니다" % (hour, private_class, prediction))
```

단순 선형 회귀: 은닉층 필요 없음 → Dense(1, linear)

다중 선형 회귀: 은닉층 필요 없음 → Dense(1, linear) (input_dim=변수 개수)

비선형 회귀(곡선/복잡한 패턴): 은닉층 필요 → Dense(..., relu) 여러 개 + 마지막 Dense(1, linear)

# 로지스틱 회귀 모델: 참 거짓 판단

```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

x = np.array([2, 4, 6, 8, 10, 12, 14]) # 공부 시간
y = np.array([0, 0, 0, 1, 1, 1, 1]) # 0은 불합격 1은 합격

model = Sequential() # 모델 생성
model.add(Dense(1, input_dim=1, activation='sigmoid'))

model.compile(optimizer='sgd' ,loss='binary_crossentropy')
model.fit(x, y, epochs=1000)


#그래프로 확인해 봅니다.

plt.scatter(x, y)
plt.plot(x, model.predict(x),'r')
plt.show()
```
